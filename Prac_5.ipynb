```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
path = '/content/drive/MyDrive/ML_lab/banking - banking (1).csv'
df=pd.read_csv(path)
df.head()
```

```python
X = df.drop(columns=['y'])
y = df['y']
```

```python
X.head()
```

```python
print(X)
```

```python
#X = pd.get_dummies(X)
```

```python
X = pd.get_dummies(X, dtype = int)
```

```python
X.head()
```

```python
X.shape
```

```python
X.columns
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test,y_train,y_test = train_test_split(X, y, train_size=0.7, test_size= 0.3,random_state=42)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_classifier = RandomForestClassifier(n_estimators=100, n_jobs=-1, max_depth=5,oob_score=True,random_state=42)
rf_classifier.fit(X_train,y_train)
```

```python
y_pred = rf_classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
print('Confusion Matrix:')
print(confusion_matrix(y_test, y_pred))
```

```python
print('Accuracy Score:')
print(accuracy_score(y_test, y_pred))
```

```python
rf_classifier.oob_score_
```

```python
print('Classification Report:')
print(classification_report(y_test, y_pred))
```

**Hypertunning Parameter**

```python
rf = RandomForestClassifier(random_state=42, n_jobs=-1)
params= {
    'max_depth': [2,3,5,10,20],
    'min_samples_leaf': [5,10,20,50,100,200],
    'n_estimators': [10,25,30,50,100,200]
}
```

```python
from sklearn.model_selection import GridSearchCV
grid_search = GridSearchCV(estimator=rf, param_grid=params, cv=4, n_jobs=-1, verbose=1, scoring="accuracy")
grid_search = grid_search.fit(X_train, y_train)
```

```python
grid_search.best_score_
```

```python
print(grid_search.best_params_)
```
