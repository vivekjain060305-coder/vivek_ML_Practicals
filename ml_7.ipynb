```python
median_exam_score = df['exam_score'].median()
df['exam_performance_category'] = df['exam_score'].apply(lambda x: 'high_performance' if x >= median_exam_score else 'low_performance')

X = df.drop(columns=['exam_score', 'exam_performance_category', 'student_id'])
y = df['exam_performance_category']
```

```python
X.head()
```

```python
X = pd.get_dummies(X, dtype=int)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test = train_test_split(X,y,train_size=0.7,random_state=42)
```

```python
X_train.shape,X_test.shape
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler  = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
svc=SVC()
svc.fit(X_train,y_train)
y_pred = svc.predict(X_test)
print('Accuracy Score')
print(accuracy_score(y_pred,y_test))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
cm=confusion_matrix(y_pred,y_test)
print(confusion_matrix(y_pred,y_test))
```

```python
svc=SVC(C=100.0)
svc.fit(X_train,y_train)
y_pred=svc.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
svc=SVC(C=1000.0)
svc.fit(X_train,y_train)
y_pred=svc.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
liner_svc=SVC(kernel='linear',C=1.0)
liner_svc.fit(X_train,y_train)
y_pred=liner_svc.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
liner_svc=SVC(kernel='linear',C=1.0)
liner_svc.fit(X_train,y_train)
y_pred=liner_svc.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
liner_svc=SVC(kernel='linear',C=1000.0)
liner_svc.fit(X_train,y_train)
y_pred=liner_svc.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
liner_svc=SVC(kernel='poly',C=1.0)
liner_svc.fit(X_train,y_train)
y_pred=liner_svc.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
liner_svc=SVC(kernel='sigmoid',C=1.0)
liner_svc.fit(X_train,y_train)
y_pred=liner_svc.predict(X_test)
accuracy_score(y_test,y_pred)
```
